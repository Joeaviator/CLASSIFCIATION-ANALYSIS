```python
from sklearn.linear_model import SGDClassifier
from sklearn.datasets import make_blobs
```

```python
#Generate the data
x,y=make_blobs(
    n_samples=1000,
    centers=2,
    random_state=0,
    cluster_std=0.5
)
```

```python
clf=SGDClassifier(loss='hinge', alpha=0.01,max_iter=1000)
```

```python
clf.fit(x,y)
```

```python
#create a grid
xx=np.linspace(-1,5,10)
yy=np.linspace(-1,5,10)
x1,x2=np.meshgrid(xx,yy)
```

```python
#decision function
z=np.empty(x1.shape)
for (i,j), current_x in np.ndenumerate(x1):
  current_y = x2[i,j]
  p=clf.decision_function([[current_x, current_y]])
  z[i,j]=p[0]
```

```python
levels=[-1.0,0.0,1.0]
```

```python
#plot the seperation
plt.contour(x1,x2,z,levels,colors='green',linestyles="--")
plt.scatter(x[:,0],x[:,1],c=y)
plt.show()
```
